# Amazon-Style E-Commerce Sales Analytics
## 01 — Data Cleaning

### Objective

The objective of this notebook is to inspect, validate, and clean the Amazon-style e-commerce dataset before performing exploratory data analysis and business analysis.

The dataset contains order, customer, product, pricing, discount, payment, geographic, and seller information.

### Key Data Cleaning Tasks

- Inspect dataset structure and dimensions
- Identify missing values
- Identify duplicate records
- Validate data types
- Check categorical consistency
- Validate numerical values
- Validate dates
- Check financial calculations
- Identify potential data-quality issues
- Create a clean dataset for further analysis

```python
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

numeric_columns
```

```python
df[numeric_columns].describe().T
```

```python
negative_values = {}

for column in numeric_columns:
    count = (df[column] < 0).sum()
    negative_values[column] = count

negative_values
```

```python
print("Minimum Quantity:", df["Quantity"].min())
print("Maximum Quantity:", df["Quantity"].max())

print(
    "Zero Quantity:",
    (df["Quantity"] == 0).sum()
)

print(
    "Negative Quantity:",
    (df["Quantity"] < 0).sum()
)
```

```python
price_columns = [
    "UnitPrice",
    "Discount",
    "Tax",
    "ShippingCost",
    "TotalAmount"
]

df[price_columns].describe().T
```

```python
for column in price_columns:
    print(
        column,
        "negative values:",
        (df[column] < 0).sum()
    )
```

```python
print("Minimum discount:", df["Discount"].min())
print("Maximum discount:", df["Discount"].max())
```

```python
df["Discount"].value_counts().sort_index().head(20)
```

```python
df["CalculatedTotal"] = (
    df["Quantity"] *
    df["UnitPrice"] *
    (1 - df["Discount"]) +
    df["Tax"] +
    df["ShippingCost"]
)
```

```python
df["AmountDifference"] = (
    df["TotalAmount"] -
    df["CalculatedTotal"]
)

df["AmountDifference"].describe()
```

```python
id_columns = [
    "OrderID",
    "CustomerID",
    "ProductID",
    "SellerID"
]

categorical_analysis_columns = [
    col for col in categorical_columns
    if col not in id_columns
]

for column in categorical_analysis_columns:
    print("\n" + "=" * 60)
    print(column)
    print("Unique values:", df[column].nunique())
    print(df[column].value_counts().head(15))
```

```python
print("OrderID format examples:")
print(df["OrderID"].head())

print("\nCustomerID format examples:")
print(df["CustomerID"].head())

print("\nProductID format examples:")
print(df["ProductID"].head())

print("\nSellerID format examples:")
print(df["SellerID"].head())
```

```python
id_columns = [
    "OrderID",
    "CustomerID",
    "ProductID",
    "SellerID"
]

for column in id_columns:
    print(
        column,
        "missing:", df[column].isna().sum(),
        "| unique:", df[column].nunique()
    )
```

```python
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    leading_trailing_spaces = (
        df[column].astype(str) != df[column].astype(str).str.strip()
    ).sum()

    print(
        f"{column}: {leading_trailing_spaces} values "
        "with leading/trailing spaces"
    )
```

```python
for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
important_categories = [
    "Category",
    "Brand",
    "PaymentMethod",
    "OrderStatus",
    "Country"
]

for column in important_categories:
    print(f"\n{column}")
    print(df[column].value_counts())
```

```python
geo_check = (
    df.groupby("Country")["State"]
    .nunique()
    .sort_values(ascending=False)
)

geo_check
```

```python
print("Tax < 0:",
      (df["Tax"] < 0).sum())
```

```python
quality_summary = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows",
        "Duplicate OrderIDs",
        "Invalid Dates",
        "Future Dates",
        "Negative Quantity",
        "Negative UnitPrice",
        "Negative Discount",
        "Negative Tax",
        "Negative ShippingCost"
    ],
    "Result": [
        len(df),
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["OrderID"].duplicated().sum(),
        df["OrderDate"].isna().sum(),
        (df["OrderDate"] > pd.Timestamp.today()).sum(),
        (df["Quantity"] < 0).sum(),
        (df["UnitPrice"] < 0).sum(),
        (df["Discount"] < 0).sum(),
        (df["Tax"] < 0).sum(),
        (df["ShippingCost"] < 0).sum()
    ]
})

quality_summary
```

```python
df.drop(
    columns=["CalculatedTotal", "AmountDifference"],
    inplace=True
)
```

```python
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month
df["MonthName"] = df["OrderDate"].dt.month_name()
df["Quarter"] = df["OrderDate"].dt.quarter
df["YearMonth"] = df["OrderDate"].dt.to_period("M").astype(str)
```

```python
output_path = "C:/Users/HP/OneDrive/Desktop/Projects/Amazon E-Commerce Sales Performance & Customer Analytics/Data/Amazon_Cleaned.csv"

df.to_csv(
    output_path,
    index=False
)

print("Cleaned dataset saved successfully.")
print("Shape:", df.shape)
```
